# 01 – City Boundary
Downloads the city boundary from OpenStreetMap and converts it to a meter-based
coordinate system (UTM zone 17N) so areas and distances can be measured accurately.

```python
import osmnx as ox

# Change this one setting later to run the project on other cities
CITY = "Toronto, Ontario, Canada"

# Download the city boundary from OpenStreetMap
city = ox.geocode_to_gdf(CITY)
print("Original CRS:", city.crs)

# Convert to a meter-based system so areas and distances are real
city_utm = city.to_crs(epsg=32617)   # UTM zone 17N (covers Toronto)
print("Area:", round(city_utm.area.iloc[0] / 1e6), "km²")

city_utm.plot(figsize=(8, 8))
```

```python
# Save the boundary so later notebooks can load it without re-downloading
city_utm.to_file("../data/processed/city_boundary.gpkg", driver="GPKG")
print("Saved")
```
